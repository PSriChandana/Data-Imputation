````python
import pandas as pd
import numpy as np
````

````python
data=pd.read_csv("GTPvar.csv",index_col=0)
````

````python
data.isnull().sum(axis=1)
#Aggregating the number of null values per row, for any column,
#because that gives us the number of missing values , so the number of
# missing variables for each row
````

````python
data["NApresent"]=data.isnull().sum(axis=1)
````

````python
data["NApresent"].value_counts()
#this gives us the number of missing values
#with respect to how many rows
````

````python
#now we filter out the rows whose number of NApresent is 0
df=data[data.NApresent==0]
#because the NApresent doesnt give any further information
df=df.drop("NApresent",axis=1)
````

````python
#Converting the pandas DataFrame to Numpy array
df_mat=df.to_numpy()
print(df_mat)
````

````python
#RANK OF MATRIX
np.linalg.matrix_rank(df_mat)
#Rank=5,rank is full, so no null space
#this shows that there is no relationship between the variables
#matrix_rank uses SVD function,
````

````python
#to use SVD decomposition-input is a matrix
#returns -1. Unitary array, 2. vector of all singular values
#3. another unitary matrix
v,s,u=np.linalg.svd(df_mat)
#v is dimension of 5,5 u is dimention of 58,58
#singular values are square roots of eigen values
````

Q. How and Why do you recompute the eigen Values?

The eigenvalues (or singular values) are very close to zero but not exactly zero, so the matrix is mathematically full rank. However, for numerical computations, a tolerance is chosen. Eigenvalues below this tolerance are treated as zero, giving an effective (numerical) rank that is smaller than the mathematical rank.

````python
#new rank is 2
#set Tolerance
tol=1e-8
#Removing columns that are lesser than the tolerance
rank=min(df_mat.shape)-np.abs(s)[::-1].searchsorted(tol)
#Choosing the null space relation
#Whats this??
A=v[:,rank:]
A=A.T
print(A)
````
